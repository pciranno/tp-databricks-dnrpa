# DDL Silver - Definiciones de Estructura

**Propósito**: Define la infraestructura de la capa Silver (esquema y tablas).

**Ejecutar**: Solo la primera vez o cuando cambie la estructura de las tablas.

**Notas**:
* Las tablas Silver almacenan datos limpios y normalizados
* Se usan tipos de datos apropiados (DATE, TIMESTAMP, STRING)
* El notebook ETL `02_Silver_dnrpa` se encarga de TRANSFORMAR y CARGAR los datos

```python
%sql
-- Crear el esquema Silver si no existe
CREATE SCHEMA IF NOT EXISTS workspace.tp_dnrpa_silver
COMMENT 'Capa Silver - Datos limpios y normalizados del DNRPA';
```

```python
%sql
-- Crear tabla silver_transferencias con esquema explícito y tipos correctos
CREATE TABLE IF NOT EXISTS workspace.tp_dnrpa_silver.silver_transferencias (
  tramite_fecha DATE COMMENT 'Fecha del trámite',
  fecha_inscripcion_inicial DATE COMMENT 'Fecha de inscripción inicial',
  registro_seccional_codigo STRING COMMENT 'Código del registro seccional (FK a dim_geografia)',
  registro_seccional_descripcion STRING COMMENT 'Descripción del registro seccional',
  registro_seccional_provincia STRING COMMENT 'Provincia del registro',
  automotor_origen STRING COMMENT 'Origen del vehículo',
  automotor_anio_modelo STRING COMMENT 'Año modelo del vehículo',
  automotor_tipo_codigo STRING COMMENT 'Código de tipo de vehículo (FK a dim_tipo_vehiculo)',
  automotor_tipo_descripcion STRING COMMENT 'Descripción del tipo',
  automotor_marca_codigo STRING COMMENT 'Código de marca (FK a dim_marca)',
  automotor_marca_descripcion STRING COMMENT 'Nombre de la marca',
  automotor_modelo_codigo STRING COMMENT 'Código de modelo (FK a dim_modelo)',
  automotor_modelo_descripcion STRING COMMENT 'Nombre del modelo',
  automotor_uso_codigo STRING COMMENT 'Código de uso del vehículo',
  automotor_uso_descripcion STRING COMMENT 'Descripción del uso',
  titular_tipo_persona STRING COMMENT 'Tipo de persona (física/jurídica)',
  titular_domicilio_localidad STRING COMMENT 'Localidad del titular',
  titular_domicilio_provincia STRING COMMENT 'Provincia del titular',
  titular_pais_nacimiento STRING COMMENT 'País de nacimiento',
  titular_porcentaje_titularidad STRING COMMENT 'Porcentaje de titularidad',
  titular_domicilio_provincia_id STRING COMMENT 'ID de provincia',
  tramite STRING COMMENT 'Tipo de trámite normalizado (ej: TRANSFERENCIA)',
  id_tramite STRING COMMENT 'Clave sintética única (PK)',
  silver_ingestion_timestamp TIMESTAMP COMMENT 'Timestamp de ingesta a Silver'
) USING DELTA
COMMENT 'Tabla Silver: transferencias vehiculares limpias y normalizadas';
```
